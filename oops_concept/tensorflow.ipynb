```python
class TensorShowcase:
    def __init__(self):
        """Initialize the class with various tensor ranks."""
        # Rank 0: scalar
        self.rank_0 = tf.constant(4)

        # Rank 1: Vector
        self.rank_1 = tf.constant([1,2,3])

        # Rank 2: Matrix (3x3)
        self.rank_2 = tf.constant([[4,5,6], [7,8,9]], dtype=tf.int32)

        # Rank 3: 3D Tensor (2x2x2)
        self.rank_3 = tf.constant([
                                [[1,2], [3,4]],
                                [[5,6], [7,8]]
                                ], dtype=tf.int32)



    def display_shapes(self):
        """Method to print the shapes of all stored tensors."""
        print(f"Rank 0 Shape: {self.rank_0.shape} (Scalar)")
        print(f"Rank 1 Shape: {self.rank_1.shape} (Vector)")
        print(f"Rank 2 Shape: {self.rank_2.shape} (Tensor)")
        print(f"Rank 3 Shape: {self.rank_3.shape} (Tensor)")

    def create_tensor(self, rows, cols):
        shape=(rows, cols)

        random_tensor = tf.random.normal(shape=shape, mean=0, stddev=1)
        return random_tensor


    def zeros_create(self, rows, cols):
        shape=(rows, cols)
        zeros_tensor = tf.zeros(shape=shape, dtype=tf.float32)
        return zeros_tensor

    def ones_create(self, rows, cols):
        shape=(rows, cols)
        ones_tensor = tf.ones(shape=shape, dtype=tf.float32)
        return ones_tensor

    # Random uniform distribution between 0 and 10
    def create_uniform(self, rows, cols):
        shape=(rows, cols)
        uniform_tensor = tf.random.uniform(shape=shape, minval=0, maxval=1)
        return uniform_tensor


tensor=TensorShowcase()
tensor.display_shapes()
tensor.create_tensor(5,3)
tensor.zeros_create(5,3)
tensor.ones_create(5,3)
tensor.create_uniform(5,3)
```

```python
# new class
class TensorOperations:
    def __init__(self):
        self.a=tf.constant([[1,2,3],[4,5,6],[7,8,9]])
        self.b=tf.constant([[1,2,3],[4,5,6],[7,8,9]])

    def addition(self):
        return self.a + self.b

    def subtraction(self):
        return self.a - self.b

    def multiplication(self):
        return self.a * self.b

    def division(self):
        return self.a / self.b


operations=TensorOperations()
operations.addition()
operations.subtraction()
operations.multiplication()
operations.division()
```

```python
# convert numpy to tensor and vise versa
import numpy as np

class ConvertNumpyToTensor:
    def __init__(self):
        self.a=np.array([1,2,3])
        self.b=tf.constant([4,5,6])

    # Convert NumPy array to PyTorch Tensor
    def convert_tensor(self):
        tensor_from_numpy=tf.convert_to_tensor(self.a, dtype=tf.float32)
        return tensor_from_numpy

    def convert_to_numpy(self):
        tensor_to_numpy=np.array(self.b)
        return tensor_to_numpy


converter=ConvertNumpyToTensor()
converter.convert_tensor()
converter.convert_to_numpy()
```

```python
class Calculation(TensorShowcase):
    def __init__(self):
        super().__init__()
        # Rank 4: 3D Tensor (2x2x2)
        self.rank_4 = tf.constant([
                                [[1,2], [3,4]],
                                [[5,6], [7,8]]
                                ], dtype=tf.int32)

    def addition(self):
        return self.rank_3 + self.rank_4
    def subtraction(self):
        return self.rank_3 - self.rank_4
    def multiplication(self):
        return self.rank_3 * self.rank_4

    @staticmethod
    def matrix_addition_example(self):
        c=tf.constant([[1,2,3],[4,5,6],[7,8,9]])
        d=tf.constant([[1,2,3],[4,5,6],[7,8,9]])
        return tf.add(c,d)


```

```python
import tensorflow as tf

class TensorShowcase:
    def __init__(self):
        # Rank 3 tensor (2x2x2)
        self.rank_3 = tf.constant([
            [[1, 1], [1, 1]],
            [[2, 2], [2, 2]]
        ], dtype=tf.int32)

class Calculation(TensorShowcase):
    def __init__(self):
        super().__init__()
        # Rank 4 tensor (2x2x2)
        self.rank_4 = tf.constant([
            [[1, 2], [3, 4]],
            [[5, 6], [7, 8]]
        ], dtype=tf.int32)

    # Operations using class attributes
    def addition(self):
        return self.rank_3 + self.rank_4

    def subtraction(self):
        return self.rank_3 - self.rank_4

    def multiplication(self):
        return self.rank_3 * self.rank_4

    # Independent matrix addition example
    @staticmethod
    def matrix_addition_example():
        a = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
        b = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
        return tf.add(a, b)
```

```python
Calculation.matrix_addition_example()
print(Calculation.matrix_addition_example())
```

```python
class Series:
    def __init__(self,num):
        self.num=num

    def add_num(self):
        return self.num*(self.num+1)/2

    def add_square(self):
        sum=0
        for i in range(1,self.num):
            sum=sum+i**2
            print(sum)
        return sum

    def add_qube(self):
        sum=0
        for i in range(1,self.num):
            sum=sum+i**3
        return sum
```

```python
series=Series(3)
print(series.add_num())
print(series.add_square())
print(series.add_qube())
```

```python
class List:
    def __init__(self, li):
        self.li=li

    def iter_list(self):
        bikes=[]
        for bike in self.li:
            bikes.append(bike)
        return bikes

    def items(self):
        for item in self.li:
            print(item)
    def squres(self):
        squre=[]
        for item in self.li:
            squre.append(item**2)
        return squre
```

```python
lis=List([1,2,3])
print(lis.iter_list())
print(lis.items())
print(lis.squres())
```

```python
class Numpy:
    def __init__(self,arr):
        self.arr=arr
    def add(self):
        return self.arr+self.arr
    def sub(self):
        return self.arr-self.arr
    def multiply(self):
        return self.arr*self.arr
```

```python
arr=Numpy(np.array([1,2,3]))
print(arr.add())
print(arr.sub())
print(arr.multiply())
```

```python
class IndexArray(Numpy):
    def __init__(self,arr):
        super().__init__(arr)

    def index(self):
        mul=self.multiply()
        return mul[0]

    def addIndex(self):
        mul=self.add()
        return mul[2]
```

```python
idx=IndexArray(np.array([1,2,3]))
print(idx.index())
print(idx.addIndex())
```

```python
class scalarIndex(IndexArray):
    def __init__(self,arr):
        super().__init__(arr)

    def addNum(self):
        num=self.add()
        return num[0]+6
    def addd(self):
        num=self.addIndex()
        return num
```

```python
scalar=scalarIndex(np.array([1,2,3]))
print(scalar.addNum())
print(scalar.addd())
```
